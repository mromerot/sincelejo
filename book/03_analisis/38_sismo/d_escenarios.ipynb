## Qué decisión informa

Un diagnóstico multiamenaza debe contemplar eventos poco frecuentes con daños potencialmente importantes. La evaluación sísmica necesita movimiento del suelo y tipologías estructurales, no indicadores climáticos.

Las perturbaciones son pruebas analíticas, no pronósticos ni proyecciones SSP. Para una evaluación futura se necesitan ensambles climáticos y escenarios de exposición coherentes.
