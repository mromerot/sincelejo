## Supuestos

Se multiplica la precipitación diaria por un factor uniforme, conservando la secuencia de días. Esto explora cambios en umbrales, pero no añade tormentas ni prolonga de forma realista estaciones secas. El cambio de lluvia no equivale al mismo cambio porcentual de inundación, falla de ladera o recarga. Tampoco modifica la ráfaga de viento.

```python
filas=[]
for factor in [.8,.9,1.,1.1,1.2]:
    mod=clima.copy(); mod['PRECTOTCORR']=mod.PRECTOTCORR*factor
    a=indicadores(mod).loc[1991:2020]
    filas.append({'factor_lluvia':factor,'lluvia_mm_anio':a.p_total.mean(),'Rx1day_medio_mm':a.rx1day.mean(),'R20_dias_anio':a.r20.mean(),'CDD_medio_dias':a.cdd.mean()})
display(pd.DataFrame(filas).round(2))
```

## Qué decisión informa

La decisión inicial es identificar los puntos donde la escorrentía supera la capacidad del cauce o drenaje y dónde una falla corta rutas o servicios. La lluvia regional demuestra variabilidad; la evidencia de intervención municipal justifica inspección, sin delimitar una mancha de amenaza.

Las perturbaciones son pruebas analíticas, no pronósticos ni proyecciones SSP. Para una evaluación futura se necesitan ensambles climáticos y escenarios de exposición coherentes. La prioridad es probar si las medidas siguen siendo útiles bajo supuestos alternativos, incluyendo sus costos de mantenimiento.
