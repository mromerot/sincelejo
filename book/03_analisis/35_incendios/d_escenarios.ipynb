## Supuestos

El percentil térmico permanece fijado en la referencia. Se explora calor y lluvia conjuntamente; el indicador no incluye ignición ni combustible, por lo que no estima número de incendios o área quemada.

```python
p90=clima.loc['1991':'2020'].groupby(clima.loc['1991':'2020'].index.month).T2M_MAX.quantile(.9)
b=clima.loc['1991':'2020']; threshold=b.index.month.map(p90).to_numpy()
filas=[]
for dt in [0,1,2]:
    for fp in [.8,1.]:
        flag=((b.T2M_MAX+dt)>threshold)&((b.PRECTOTCORR*fp)<1)
        filas.append({'delta_C':dt,'factor_lluvia':fp,'dias_secos_calidos_anio':flag.groupby(b.index.year).sum().mean()})
display(pd.DataFrame(filas).round(2))
```
