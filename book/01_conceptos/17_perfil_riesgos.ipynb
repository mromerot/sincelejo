## Evidencia local

## Perfil meteorológico

Comparar una referencia de 30 años con cinco años recientes es descriptivo; el bloque reciente no es una nueva normal.

```python
comparacion=pd.DataFrame({'1991–2020':anual.loc[1991:2020].mean(),'2021–2025':anual.loc[2021:2025].mean()})
display(comparacion.round(2))
```

## Lectura territorial

Las amenazas localizadas necesitan cartografía; las dependencias de agua y energía necesitan redes; salud necesita series y denominadores. El resultado es una agenda de análisis, no una clasificación oficial de riesgo.
